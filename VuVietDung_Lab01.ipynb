# Ghi chú học tập: Jupyter Notebook

## Các thao tác đã học
- Tạo notebook và chạy code Python
- Thêm cell mới (phím `a`, `b`)
- Chuyển cell sang Markdown (phím `m`)
- Đổi tên, lưu và tải notebook

Tài liệu tham khảo: [Markdown Guide - Basic Syntax](https://www.markdownguide.org/basic-syntax/)

| Phím tắt | Chức năng |

| b | Thêm cell phía dưới |
